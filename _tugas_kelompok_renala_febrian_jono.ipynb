```python
# ============================================================
# TUGAS AKUISISI DAN MANAJEMEN DATA
# Tema: Peta & Ulasan Warung Kuliner Denpasar
# Analisis Karakteristik Data dan Estimasi Storage
# ============================================================

import pandas as pd
import os

# 1. DATA ASUMSI BULAN PERTAMA
data = {
    "Jenis Data": [
        "Ulasan pelanggan",
        "Foto makanan dan warung",
        "Titik lokasi warung",
        "Catatan menu"
    ],
    "Format": [
        "TXT / CSV / JSON",
        "JPG / JPEG / PNG",
        "GeoJSON",
        "CSV / JSON"
    ],
    "Jumlah Unit": [1000, 200, 50, 300],
    "Ukuran Rata-rata (KB)": [1.5, 850, 0.3, 2],
    "Atribut Teknis": [
        "Encoding UTF-8, jumlah karakter, bahasa",
        "Resolusi, piksel, channel warna, kompresi",
        "Point, longitude, latitude, CRS",
        "Nama kolom, tipe data, harga numerik"
    ],
    "Variasi": [
        "Bahasa, singkatan, emoji, typo",
        "Resolusi, kualitas, format gambar",
        "Koordinat dan kategori lokasi",
        "Nama menu, harga, kategori"
    ],
    "Metadata": [
        "ID ulasan, tanggal, ID warung, rating",
        "Nama file, resolusi, tanggal, ID warung",
        "Nama warung, longitude, latitude, CRS",
        "ID menu, ID warung, nama menu, harga"
    ]
}

df = pd.DataFrame(data)

print("TABEL ANALISIS KARAKTERISTIK DATA")
display(df)

# 2. MENGHITUNG TOTAL DATA BULAN PERTAMA

df["Total Ukuran (KB)"] = (
    df["Jumlah Unit"] * df["Ukuran Rata-rata (KB)"]
)

total_kb = df["Total Ukuran (KB)"].sum()
total_mb = total_kb / 1024
total_gb = total_mb / 1024

print("\nESTIMASI STORAGE BULAN PERTAMA")
display(df[[
    "Jenis Data",
    "Jumlah Unit",
    "Ukuran Rata-rata (KB)",
    "Total Ukuran (KB)"
]])

print(f"Total data       : {total_kb:,.2f} KB")
print(f"Total data       : {total_mb:,.2f} MB")
print(f"Total data       : {total_gb:.4f} GB")

# 3. ESTIMASI STORAGE SELAMA 12 BULAN
# Asumsi data BARU bertambah 10% setiap bulan

pertumbuhan = 0.10
jumlah_bulan = 12
cadangan = 0.30

faktor = (
    (1 + pertumbuhan) ** jumlah_bulan - 1
) / pertumbuhan

total_tahunan_kb = total_kb * faktor
cadangan_kb = total_tahunan_kb * cadangan
total_akhir_kb = total_tahunan_kb + cadangan_kb

print("\nESTIMASI STORAGE SELAMA 12 BULAN")
print(f"Pertumbuhan per bulan : {pertumbuhan:.0%}")
print(f"Faktor akumulasi      : {faktor:.4f}")
print(f"Total 12 bulan        : {total_tahunan_kb / 1024**2:.4f} GB")
print(f"Cadangan 30%          : {cadangan_kb / 1024**2:.4f} GB")
print(f"Total kebutuhan       : {total_akhir_kb / 1024**2:.4f} GB")

# 4. TABEL PERTUMBUHAN PER BULAN

hasil_bulanan = []
for bulan in range(1, jumlah_bulan + 1):
    volume = total_kb * (1 + pertumbuhan) ** (bulan - 1)

    hasil_bulanan.append({
        "Bulan": bulan,
        "Data Baru (MB)": volume / 1024,
        "Akumulasi (MB)": sum(
            total_kb * (1 + pertumbuhan) ** (i - 1)
            for i in range(1, bulan + 1)
        ) / 1024
    })

df_bulanan = pd.DataFrame(hasil_bulanan)

print("\nPERTUMBUHAN DATA BARU SETIAP BULAN")
display(df_bulanan.round(2))

# 5. MENYIMPAN HASIL KE CSV

os.makedirs("hasil_analisis", exist_ok=True)

df.to_csv(
    "hasil_analisis/karakteristik_data.csv",
    index=False,
    encoding="utf-8-sig"
)

df_bulanan.to_csv(
    "hasil_analisis/pertumbuhan_storage.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nFile berhasil dibuat:")
print("- hasil_analisis/karakteristik_data.csv")
print("- hasil_analisis/pertumbuhan_storage.csv")
```
